# Fraud detection - exploratory data analysis

```python
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
DATASET_DIR = Path("../data/raw/creditcard.csv")

df = pd.read_csv(DATASET_DIR)
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
TARGET = "Class"
```

```python
counts = df[TARGET].value_counts()
```

```python
plt.figure(dpi=100,figsize=(5,5))
plt.bar(["Non-fraud", "Fraud"], counts)
plt.show()
```

```python
numerical = df.select_dtypes(include='float64').columns.to_list()
```

```python
plt.figure(dpi=100, figsize=(20,20))
sns.heatmap(df[numerical + [TARGET]].corr(numeric_only=True), annot=True)
plt.show()
```

```python
for x in numerical:
    plt.figure(dpi=77, figsize=(7,7))
    plt.title(f"Distribution of {x}")
    plt.hist(df[x], bins="auto")
    plt.show()
```
